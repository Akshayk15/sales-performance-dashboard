```python
import pandas as pd

df = pd.read_csv("superstore_final_clean.csv")
print(df.shape)
print(df.head())
```

```python

import sqlite3

conn = sqlite3.connect("superstore.db")
df.to_sql("orders", conn, if_exists="replace", index=False)
conn.close()
print("Loaded into superstore.db")
```

```python
conn = sqlite3.connect("superstore.db")
```

```python
import pandas as pd

query1 = """
SELECT "Product Name", SUM(Profit) AS total_profit
FROM orders
GROUP BY "Product Name"
ORDER BY total_profit DESC
LIMIT 10;
"""
top_products = pd.read_sql(query1, conn)
print(top_products)
```

```python
query2 = """
SELECT strftime('%Y-%m', "Order Date") AS month, SUM(Sales) AS revenue
FROM orders
GROUP BY month
ORDER BY month;
"""
monthly_revenue = pd.read_sql(query2, conn)
print(monthly_revenue)
```

```python
query3 = """
SELECT Segment, COUNT(DISTINCT "Customer Name") AS num_customers,
       SUM(Sales) AS total_sales, AVG(Profit) AS avg_profit
FROM orders
GROUP BY Segment
ORDER BY total_sales DESC;
"""
segment_analysis = pd.read_sql(query3, conn)
print(segment_analysis)
```

```python
conn = sqlite3.connect("superstore.db")
query4 = """
SELECT "Product Name", SUM(Sales) AS total_sales, SUM(Profit) AS total_profit
FROM orders
GROUP BY "Product Name"
ORDER BY total_sales DESC
LIMIT 10;
"""
top_by_sales = pd.read_sql(query4, conn)
print(top_by_sales)
conn.close()
```

```python
conn = sqlite3.connect("superstore.db")
query5 = """
SELECT "Product Name", AVG(Discount) AS avg_discount, SUM(Profit) AS total_profit
FROM orders
WHERE "Product Name" IN ('Cisco TelePresence System EX90 Videoconferencing System', 'GBC DocuBind P400 Electric Binding System', 'HON 5400 Series Task Chairs for Big and Tall')
GROUP BY "Product Name";
"""
print(pd.read_sql(query5, conn))
conn.close()
```
