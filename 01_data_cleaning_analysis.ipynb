```python
import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

# This automatically detects and parses BOTH formats correctly
df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed", dayfirst=False)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="mixed", dayfirst=False)

print(df["Order Date"].head(20))
print(df.dtypes)  # confirm both columns say datetime64
```

```python
df.to_csv("superstore_clean_dates.csv", index=False)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna(subset=["Sales", "Profit", "Order Date"])
print("Rows remaining:", len(df))
```

```python
print("Duplicate rows found:", df.duplicated().sum())
df = df.drop_duplicates()
print("Rows after removing duplicates:", len(df))
```

```python
sales_by_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(sales_by_region)
```

```python
sales_by_category = df.groupby("Category")[["Sales", "Profit"]].sum()
print(sales_by_category)
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")
sales_by_month = df.groupby("Month")["Sales"].sum()
print(sales_by_month)
```

```python
df.to_csv("superstore_final_clean.csv", index=False)
print("Saved successfully")
```
